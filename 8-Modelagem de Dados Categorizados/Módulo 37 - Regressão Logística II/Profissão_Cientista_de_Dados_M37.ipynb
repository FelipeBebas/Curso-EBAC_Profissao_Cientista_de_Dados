# Tarefa I

1. Qual a principal diferença entre um *behaviour score* e um *credit score*?

> Credit score avalia risco antes da concessão de crédito. Behaviour score avalia risco após a concessão, baseado no comportamento atual do cliente.

2. Por que é tão comum se estabelecer janelas de desempenho de 12 meses, e 12 safras para o desenvolvimento de modelos?

> Capturam comportamento anual completo e variações sazonais. As safras testam o modelo em diferentes períodos, garantindo robustez.

3. Qual a diferença entre amostra *out of time* e amostra *out of sample*?

> - Out of time: dados de período futuro.
- Out of sample: dados separados aleatoriamente do mesmo período, não usados no treino.


4. Se os dados da variável resposta estão sistematicamente corrompidos para duas das safras de desenvolvimento, o que você faria?

> Identificar e corrigir se possível. Caso contrário, excluir as safras e avaliar impacto, compensando com outras.
